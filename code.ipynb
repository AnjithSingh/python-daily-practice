```python
# prime or not

def prime(n):
    if n<2:
        return "NOTPRIME"
    if n == 2: 
        return 'prime'
    for i in range(2,n):
        if n%i == 0:
            return 'NOTPRIME'
    return 'PRIME'

print(prime(21))
```

```python
def fact(n):
    if n<0:
        return "not exist"
    if n==1:
        return "0"
    factorial = 1
    for i in range(1,n+1):
        factorial= factorial * i
    return factorial
        

print(fact(5))
```

```python
# fibonacci

def fibo(n):
    a = 0
    b = 1 
    
    if a == 1:
        print(a)
    else:
        print(a)
        print(b)
    for i in range(1,n+1):
        c = a+b
        a = b
        b = c 
        print(c)
    



print(fibo(7))
```

```python
# Armstrong num

def arm(n):
    a = 0
    n = str(n)
    for i in (n):
        a = a+ int(str(n)[i])**3
    if a == int(n):
        return "ARMSTRONG"
    else:
        return "NOT ARMSTRONG"

print(arm(153))
```
